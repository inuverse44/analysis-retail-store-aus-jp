```kotlin
val df2Double = df.convert("lat", "lon").toDouble()
df2Double.size()
```

Lets-Plotで地図にプロット

```kotlin
import org.jetbrains.letsPlot.*
import org.jetbrains.letsPlot.geom.*
import org.jetbrains.letsPlot.coord.*

val colesMap = df2Double.toMap()

val plot = letsPlot(colesMap) +
        geomPoint(
            color = "red",
            size = 2.0,
            alpha = 0.2
        ) {
            x = "lon"
            y = "lat"
        } +
        coordMap()

plot
```

```kotlin
/**
 * オーストラリアは
 * ・東経110~155度
 * ・南緯10~45度
 * に位置する。
 * 東経は+、西経は-。同様に北緯は+、南緯は-。
 */

val plotZoomed = letsPlot(colesMap) +
        geomPoint(color = "red", size = 1.0, alpha = 0.3) {
            x = "lon"
            y = "lat"
        } +
        coordMap(
            xlim = Pair(110.0, 155.0),
            ylim = Pair(-45.0, -10.0)
        )

plotZoomed
```
